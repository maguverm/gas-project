```python
import pandas as pd
df_check = pd.read_parquet('../data/processed/contratos_gas.parquet',
                            columns=['no_operacion', 'fecha_inicial', 'fecha_final', 
                                     'sector_consumo', 'cantidad', 'precio'])

# Cuántos registros únicos hay con cada combinación
dedup_actual = df_check.drop_duplicates(
    subset=['no_operacion', 'fecha_inicial', 'fecha_final', 'sector_consumo']
)

dedup_correcto = df_check.drop_duplicates(
    subset=['no_operacion', 'fecha_inicial', 'fecha_final', 'sector_consumo', 'cantidad']
)

print(f"Con dedup actual:   {len(dedup_actual):,} contratos")
print(f"Con dedup correcto: {len(dedup_correcto):,} contratos")
```

```python
import pandas as pd

df_dem = pd.read_csv('../data/raw/demanda_gestor_consolidado.csv', encoding='latin-1')

# Renombrar columnas
df_dem.columns = ['fecha_registro', 'tipo_demanda', 'nombre_operador', 
                  'sector_consumo', 'cantidad_entregada', 'anio', 'semestre']

# Corregir encoding
df_dem['sector_consumo'] = df_dem['sector_consumo'].str.encode('latin-1').str.decode('utf-8')
df_dem['nombre_operador'] = df_dem['nombre_operador'].str.encode('latin-1').str.decode('utf-8')

# Convertir fecha
df_dem['fecha_registro'] = pd.to_datetime(df_dem['fecha_registro'])

# Convertir categóricas
for col in ['tipo_demanda', 'nombre_operador', 'sector_consumo', 'semestre']:
    df_dem[col] = df_dem[col].astype('category')

print(df_dem.dtypes)
print(f"\nSectores: {df_dem['sector_consumo'].unique()}")

# Guardar en processed
df_dem.to_parquet('../data/processed/demanda_gestor.parquet', index=False)
print("\n✅ Parquet guardado")
```

```python
import pandas as pd

normalizar_empresas = {
    'EMGESA S.A ESP': 'EMGESA S.A. E.S.P.',
    'Gases del Cusiana SA Empresa de Servicios Públicos': 'GASES DEL CUSIANA S.A E.S.P',
    'Gases del Llano S.A. Empresa de Servicios Públicos': 'Gases del Llano S.A. E.S.P.',
    'KRONOS ENERGY S.A. E.S.P': 'KRONOS ENERGY S.A. E.S.P.',
    'PRIME TERMOVALLE S.A.S. E.S.P.': 'ENFRAGEN TERMOVALLE S.A.S. E.S.P.',
    'Prime Termoflores S.A.S. E.S.P.': 'Enfragen Termoflores S.A.S. E.S.P.',
    'Gases del Cusiana S.A.S Empresa de Servicios Públicos BIC': 'GASES DEL CUSIANA S.A E.S.P',
    'Gases del Llano S.A. Empresa de Servicios Públicos BIC': 'Gases del Llano S.A. E.S.P.',
    'KRONOS ENERGY S.A.S E.S.P': 'KRONOS ENERGY S.A. E.S.P.',
    'DEL LLANO S A': 'Gases del Llano S.A. E.S.P.',
}

# Actualizar contratos
df_cont = pd.read_parquet('../data/processed/contratos_gas.parquet')
df_cont['nombre_comprador'] = df_cont['nombre_comprador'].astype(str).replace(normalizar_empresas)
df_cont['nombre_vendedor'] = df_cont['nombre_vendedor'].astype(str).replace(normalizar_empresas)
df_cont.to_parquet('../data/processed/contratos_gas.parquet', index=False)
print(f"✅ Contratos actualizados")

# Actualizar demanda
df_dem = pd.read_parquet('../data/processed/demanda_gestor.parquet')
df_dem['nombre_operador'] = df_dem['nombre_operador'].astype(str).replace(normalizar_empresas)
df_dem.to_parquet('../data/processed/demanda_gestor.parquet', index=False)
print(f"✅ Demanda actualizada")

# Verificar
compradores = set(df_cont['nombre_comprador'].dropna().unique())
operadores = set(df_dem['nombre_operador'].dropna().unique())
print(f"\nCoinciden ahora: {len(compradores & operadores)}")
print(f"Aún sin coincidencia:")
for x in sorted(operadores - compradores):
    print(f"  {x}")
```

```python
import pandas as pd
import os

print("Leyendo CSV original...")
columnas = [
    'fecha_dia', 'cantidad', 'precio', 'sector_consumo',
    'modalidad', 'mercado', 'fecha_negociacion', 'nombre_vendedor', 
    'nombre_comprador', 'fecha_inicial', 'fecha_final', 
    'no_operacion', 'tipo_demanda'
]

df = pd.read_csv('../data/raw/Contratos 2021-2025.csv', 
                 encoding='latin-1',
                 usecols=columnas,
                 parse_dates=['fecha_dia', 'fecha_inicial', 'fecha_final'])

# Normalizar empresas
normalizar_empresas = {
    'EMGESA S.A ESP': 'EMGESA S.A. E.S.P.',
    'Gases del Cusiana SA Empresa de Servicios Públicos': 'GASES DEL CUSIANA S.A E.S.P',
    'Gases del Llano S.A. Empresa de Servicios Públicos': 'Gases del Llano S.A. E.S.P.',
    'KRONOS ENERGY S.A. E.S.P': 'KRONOS ENERGY S.A. E.S.P.',
    'PRIME TERMOVALLE S.A.S. E.S.P.': 'ENFRAGEN TERMOVALLE S.A.S. E.S.P.',
    'Prime Termoflores S.A.S. E.S.P.': 'Enfragen Termoflores S.A.S. E.S.P.',
    'Gases del Cusiana S.A.S Empresa de Servicios Públicos BIC': 'GASES DEL CUSIANA S.A E.S.P',
    'Gases del Llano S.A. Empresa de Servicios Públicos BIC': 'Gases del Llano S.A. E.S.P.',
    'KRONOS ENERGY S.A.S E.S.P': 'KRONOS ENERGY S.A. E.S.P.',
    'DEL LLANO S A': 'Gases del Llano S.A. E.S.P.',
}
df['nombre_comprador'] = df['nombre_comprador'].astype(str).replace(normalizar_empresas)
df['nombre_vendedor'] = df['nombre_vendedor'].astype(str).replace(normalizar_empresas)

for col in ['sector_consumo', 'modalidad', 'mercado',
            'nombre_vendedor', 'nombre_comprador', 'tipo_demanda']:
    df[col] = df[col].astype('category')

df.to_parquet('../data/processed/contratos_gas.parquet',
              engine='pyarrow', index=False,
              version='2.6',
              coerce_timestamps='ms',
              allow_truncated_timestamps=True)

size = os.path.getsize('../data/processed/contratos_gas.parquet') / 1024**2
print(f"✅ Contratos: {size:.1f} MB | {len(df):,} filas")
```
